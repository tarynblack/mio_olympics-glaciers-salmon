# Blue Glacier

## General time series data
_text copied from README in the data directory_

`general_data.txt` contains annual precipitation, equilibrium line altitude, length, and three mass balance series. This file was created by downloading the data [webpage](https://earthweb.ess.washington.edu/Glaciology/projects/blue_glac/tabular_data.html) as an HTML file, copying the text for the table to a new file, and find-and-replacing all tabs with commas for ease of use. There was additional manual editing to remove the parentheses and asterisks (see below), and add a column with comments containing the corresponding information.

The webpage describes the data:
> Tabulated by year below are:
> * precipitation **P** (m water equivalent until 30 September of the year shown) at a low level station near Blue Glacier (Forks, Washington);
> * equilibrium line altitude **ELA** (m above NGVD 1929, which is approximate sea level);
> * glacier length **L** (m from head of the glacier);
> * three mass balance series (m water equivalent until 30 September of the year shown):  
>   - **b'** from 1956 to 1986 was published by Armstrong (1989) based on stake measurements and probing discussed previously. Those in parentheses were estimated by Armstrong. Since 1986 we have estimated the annual balance from late season measurements of the ELA and these are marked with an asterisk (*).
>   - **b1** series is created by subtracting a single constant from each b' value so that the series is consistent with the 1957-95 mass change derived from the mapping.
>   - **b2** series is created by subtracting one constant from each value between 1957-87 and another is subtracted between 1987-95 so that the series is consistent with both the 1957-87 and 1987-95 mass changes.

```python
blue_general = pd.read_csv(DATADIR.joinpath('UW_BlueGlacier/general_data.txt'))
blue_general
```

## Mass balance data
_text copied from README in the data directory_

`massbalance_data.txt` contains the annual net mass balance, estimated error, and cumulative mass balance, as well as the equilibrium line altitude and accumulation area ratio. This file was created by downloading the data [webpage](https://earthweb.ess.washington.edu/Glaciology/projects/blue_glac/b_timeseries.htm) as an HTML file, copying the text for the table to a new file, and find-and-replacing spaces with commas column-by-column, adding commas where data was missing to maintain proper column alignment.

The webpage describes the data:
> ELA is the late season altitude of the equilibrium line in meters above the National Geodetic Vertical Datum of 1929, AAR is the fraction of the glacier area above that altitude, b_n is the net balance in water equivalent meters, sigma is its estimated standard error, and cumulative values SIGMA b_n are since the end of the 1955 ablation season.
> 
> Values for 1956-1997 are from Conway et al (1999), which adjusted b_n given by Armstrong (1989) to be consistent with measured volume changes of the glacier; it also extended the record to later years by applying to late-season observations of the ELA the relation (in terms of the AAR) obtained in earlier years when both b_n and the ELA were measured. That relation was also used to get values for 1998 and 1999.
> 
> Although large sigma are ascribed to b_n in individual years, cumulative values over selected periods (between measurements of volume change) are more closely constrained: 1.9 m over 1957-87, 2.8 m over 1987-95, 3.2 m over 1957-95. Years when b_n was measured have sigma = 0.5, others 1.0.
> 
> Values here reflect the best judgment of H. Conway and L. A. Rasmussen. The mass balance program was suspended in 1999.

```python
blue_massbal = pd.read_csv(DATADIR.joinpath('UW_BlueGlacier/massbalance_data.txt'))
blue_massbal
```

```python
### plot a variable time series for Blue Glacier
# blue_general: general time series data
# -- options: Year, P, ELA, L, b', b_2, b_1
# blue_massbal: mass balance time series data
# -- options: Year, ELA, AAR, b_n, sigma, SIGMA_b_n

data = blue_general
# data = blue_massbal
x_col = 'Year'
y_col = "L"
# y_col = "b_n"
title_str = "Annual length of Blue Glacier"
# y_str = f"{y_col}, mass balance [m w.e.]"
y_str = f"{y_col}, length [m from head]"

fig, ax = plt.subplots()
# ax.plot(t, blue_general[col])
# ax.grid(alpha=0.4)

data.plot(
    ax=ax,
    x=x_col,
    y=y_col,
    legend=False,
    xlabel=x_col,
    ylabel=y_str,
)
ax.set_title(title_str)
ax.grid(alpha=0.4)
plt.show()
```
