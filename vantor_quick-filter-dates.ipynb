```python
with open(fstr) as F:
    data = json.load(F)

df = pd.json_normalize(data, 'features')
```

```python
imgs = pd.DataFrame(
    data={
        'date': pd.to_datetime(df['properties.datetime']), 
        'id': df['properties.id'],
    },
)

imgs['year'] = imgs.date.dt.year
imgs['month'] = imgs.date.dt.month
# imgs.drop()
imgs.drop_duplicates(subset=['month', 'year'], ignore_index=True, inplace=True)
imgs.head()
```

```python
timetable = imgs.pivot(index='year', columns='month', values='id')
timetable
```
